7. Какое имя наиболее популярно среди пассажиров мужского пола?
* Charles
* Thomas
* William
* John

```python
def extract_name(full_name):
  name_without_surname = re.sub(r'^[^,]+, ', '', full_name)
  name_without_titles = re.sub(r'\b(Mr|Mrs|Miss|Master|Dr|Rev)\.?\s*', '', name_without_surname)
  first_name = name_without_titles.split()[0] if name_without_titles.split() else full_name
  first_name = re.sub(r'[\"\(\)]', '', first_name)
  return first_name
```

```python
data['first_name'] = data['Name'].apply(extract_name)
most_common_sex = data[data['Sex'] == "male"]['first_name'].mode()[0]
print(most_common_sex)
```
